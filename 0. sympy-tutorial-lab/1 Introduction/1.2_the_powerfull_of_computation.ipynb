The real power of a symbolic computation system such as SymPy is the ability to do all sorts of computations symbolically. SymPy can simplify expressions, compute derivatives, integrals, and limits, solve equations, work with matrices, and much, much more, and do it all symbolically. It includes modules for plotting, printing (like 2D pretty printed output of math formulas lATEX, or 
), code generation, physics, statistics, combinatorics, number theory, geometry, logic, and more. Here is a small sampling of the sort of symbolic power SymPy is capable of, to whet your appetite.

```python
from sympy import *
x,t,z, nu = symbols('x t z nu')
```

This will make all further examples pretty print with unicode characters.



```python
init_printing(use_unicode=True)
```

```python
diff(sin(x)*exp(x),x)
# on derive la fonction entre diff( , x) suivant x
```

```python
integrate(exp(x)*sin(x) + exp(x)*cos(x),x)
# on integre la fonction entre integrate( , x) suivant x
```

```python
diff(tan(x),x)
```

```python
diff(sin(x),x)
```

```python
diff(cos(x),x)
```

```python
integrate(sin(x**2),(x,-oo,+oo))
```

```python
f = sin(x**2)
f
```

```python
h = integrate(x**3 + 4, (x,-1,1))
h
```

```python
y = symbols("y")
M = integrate(x**3 + y**3, (x,-1,12), (y,-1,12))
int(M/1690)
```

```python
limit(sin(x)/x,x,0)
# calculer la limite de sin(x)/x comme pour Hospital
```

```python
limit(cos(x)/x,x,0)
```

```python
solve(x**2 - 4, x)
```

```python
solve(x**3 - 2*x**2 + x - 12, x)
```
